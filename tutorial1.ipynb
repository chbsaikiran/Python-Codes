# Python Data Structures: Lists, Dictionaries, Sets & Tuples
A practical tour of the most-used Python built-in collections.

---
## 1. LISTS

```python
# --- Creating lists ---
fruits = ['apple', 'banana', 'cherry']
numbers = [10, 20, 30, 40, 50]
mixed  = [1, 'hello', 3.14, True]
empty  = []
from_range = list(range(5))   # [0, 1, 2, 3, 4]
print(fruits, numbers, mixed, from_range)
```

```python
# --- Indexing & Slicing ---
print(fruits[0])      # first  -> 'apple'
print(fruits[-1])     # last   -> 'cherry'
print(numbers[1:4])   # slice  -> [20, 30, 40]
print(numbers[::2])   # every other -> [10, 30, 50]
print(numbers[::-1])  # reversed   -> [50, 40, 30, 20, 10]
```

```python
# --- Mutating a list ---
fruits[1] = 'blueberry'          # update in-place
fruits.append('mango')           # add to end
fruits.insert(1, 'avocado')      # insert at index
fruits.extend(['grape', 'kiwi']) # add multiple items
print(fruits)
```

```python
# --- Removing items ---
fruits.remove('avocado')   # remove by value
popped = fruits.pop()      # remove & return last item
del fruits[0]              # remove by index
print(fruits, '| popped:', popped)
```

```python
# --- Searching & counting ---
nums = [3, 1, 4, 1, 5, 9, 2, 6, 5]
print('index of 5 :', nums.index(5))   # first occurrence
print('count of 1 :', nums.count(1))
print('5 in list  :', 5 in nums)
```

```python
# --- Sorting & reversing ---
nums_copy = nums.copy()
nums_copy.sort()                          # sort in-place
print('sorted         :', nums_copy)
print('sorted desc    :', sorted(nums, reverse=True))  # new list
nums_copy.reverse()                       # reverse in-place
print('reversed       :', nums_copy)
```

```python
# --- List comprehensions ---
squares   = [x**2 for x in range(1, 6)]
evens     = [x for x in range(10) if x % 2 == 0]
nested    = [i*j for i in range(1,4) for j in range(1,4)]
upper     = [s.upper() for s in ['hello', 'world']]
print(squares, evens, nested, upper)
```

```python
# --- Useful built-ins on lists ---
data = [4, 7, 2, 9, 1]
print('len :', len(data))
print('min :', min(data))
print('max :', max(data))
print('sum :', sum(data))
print('zip :', list(zip([1,2,3], ['a','b','c'])))
print('enumerate sample:', list(enumerate(['x','y','z'])))
```

```python
# --- Flattening a nested list ---
nested_list = [[1,2], [3,4], [5,6]]
flat = [item for sub in nested_list for item in sub]
print('flat:', flat)

# Unpacking trick
a, *middle, z = [10, 20, 30, 40, 50]
print(f'first={a}, middle={middle}, last={z}')
```

## 2. DICTIONARIES

```python
# --- Creating dictionaries ---
person = {'name': 'Alice', 'age': 30, 'city': 'NYC'}
from_keys = dict.fromkeys(['a','b','c'], 0)
zipped    = dict(zip(['x','y'], [10, 20]))
print(person)
print(from_keys)
print(zipped)
```

```python
# --- Accessing values ---
print(person['name'])                  # KeyError if missing
print(person.get('age'))               # None if missing
print(person.get('salary', 'N/A'))     # default if missing
```

```python
# --- Adding / updating / removing ---
person['email'] = 'alice@example.com'  # add new key
person['age']   = 31                   # update existing key
person.update({'city': 'LA', 'hobby': 'chess'})  # bulk update

removed = person.pop('hobby')          # remove & return
person.popitem()                       # remove last inserted pair
print(person, '| removed:', removed)
```

```python
# --- Iterating ---
scores = {'math': 95, 'science': 88, 'english': 72}

print('keys  :', list(scores.keys()))
print('values:', list(scores.values()))
print('items :')
for subject, mark in scores.items():
    print(f'  {subject}: {mark}')
```

```python
# --- Membership & safe defaults ---
print('math in scores  :', 'math' in scores)
print('art in scores   :', 'art' in scores)

# setdefault - insert key only if absent
scores.setdefault('art', 0)
print('after setdefault:', scores)
```

```python
# --- Dictionary comprehensions ---
squared = {x: x**2 for x in range(1, 6)}
filtered = {k: v for k, v in scores.items() if v >= 88}
inverted = {v: k for k, v in {'a':1,'b':2,'c':3}.items()}
print(squared)
print(filtered)
print(inverted)
```

```python
# --- Merging dictionaries ---
d1 = {'a': 1, 'b': 2}
d2 = {'b': 99, 'c': 3}

merged_py39 = d1 | d2          # Python 3.9+ merge operator
merged_unpack = {**d1, **d2}   # works everywhere
print(merged_py39)
print(merged_unpack)
```

```python
# --- Nested dictionaries ---
employees = {
    'E01': {'name': 'Bob',   'dept': 'Engineering'},
    'E02': {'name': 'Carol', 'dept': 'Marketing'},
}
print(employees['E01']['name'])

# Auto-creating nested keys with defaultdict
from collections import defaultdict
graph = defaultdict(list)
graph['A'].append('B')
graph['A'].append('C')
print(dict(graph))
```

## 3. SETS

```python
# --- Creating sets ---
colors = {'red', 'green', 'blue'}
from_list = set([1, 2, 2, 3, 3, 4])   # duplicates removed
empty_set  = set()                     # NOT {} — that makes a dict
print(colors)
print(from_list)   # {1, 2, 3, 4}
```

```python
# --- Adding & removing ---
colors.add('yellow')           # add one element
colors.update(['pink', 'cyan'])# add multiple
colors.discard('red')          # remove if present (no error)
colors.remove('green')         # remove; raises KeyError if absent
popped = colors.pop()          # remove & return arbitrary element
print(colors, '| popped:', popped)
```

```python
# --- Set operations ---
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print('union       :', A | B)          # or A.union(B)
print('intersection:', A & B)          # or A.intersection(B)
print('difference  :', A - B)          # in A but not B
print('sym diff    :', A ^ B)          # in either but not both
```

```python
# --- In-place set operations ---
X = {1, 2, 3}
X |= {3, 4}        # union update
print('after |=  :', X)
X &= {2, 3, 4, 5}  # intersection update
print('after &=  :', X)
X -= {3}           # difference update
print('after -=  :', X)
```

```python
# --- Subset / superset checks ---
small = {1, 2}
big   = {1, 2, 3, 4}
print('small <= big (subset)  :', small <= big)
print('small <  big (proper)  :', small < big)
print('big   >= small (super) :', big >= small)
print('disjoint               :', small.isdisjoint({5, 6}))
```

```python
# --- Set comprehension ---
sq_set = {x**2 for x in range(-3, 4)}
print('square set (no dups):', sq_set)

# Practical: fast duplicate removal while preserving order
data = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
unique_ordered = list(dict.fromkeys(data))
print('unique ordered:', unique_ordered)
```

```python
# --- frozenset: immutable/hashable set ---
fs = frozenset([1, 2, 3])
print(fs)
# fs.add(4)  # would raise AttributeError
d = {fs: 'frozen key'}   # can be used as a dict key
print(d)
```

## 4. TUPLES

```python
# --- Creating tuples ---
point   = (3, 7)
rgb     = (255, 128, 0)
single  = (42,)          # trailing comma required for 1-element tuple
packed  = 1, 2, 3        # parentheses are optional
from_it = tuple('abc')   # ('a', 'b', 'c')
print(point, rgb, single, packed, from_it)
```

```python
# --- Indexing & slicing (same as list) ---
t = (10, 20, 30, 40, 50)
print(t[0])       # 10
print(t[-1])      # 50
print(t[1:4])     # (20, 30, 40)
print(t[::-1])    # reversed
```

```python
# --- Immutability ---
t = (1, 2, 3)
# t[0] = 99  # TypeError: 'tuple' object does not support item assignment

# But a tuple CAN contain mutable objects
t2 = ([1, 2], [3, 4])
t2[0].append(99)
print(t2)   # ([1, 2, 99], [3, 4])
```

```python
# --- Tuple unpacking ---
x, y = (3, 7)
r, g, b = rgb
first, *rest = (1, 2, 3, 4, 5)
*head, last  = (1, 2, 3, 4, 5)

print(f'x={x}, y={y}')
print(f'r={r}, g={g}, b={b}')
print(f'first={first}, rest={rest}')
print(f'head={head}, last={last}')

# Swap without temp variable
a, b = 5, 10
a, b = b, a
print(f'swapped: a={a}, b={b}')
```

```python
# --- Tuple methods (only 2!) ---
t = (1, 2, 3, 2, 4, 2)
print('count of 2  :', t.count(2))
print('index of 3  :', t.index(3))
```

```python
# --- Tuples as dict keys (lists can't be) ---
locations = {}
locations[(40.7128, -74.0060)] = 'New York'
locations[(51.5074, -0.1278)]  = 'London'
print(locations)

# Named tuples for readable record-like data
from collections import namedtuple
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 7)
print(f'Point: x={p.x}, y={p.y}')
print(p)          # Point(x=3, y=7)
```

```python
# --- Tuple tricks ---
# Concatenation & repetition
t1 = (1, 2) + (3, 4)
t2 = ('ha',) * 3
print(t1, t2)

# Useful in function return values
def min_max(lst):
    return min(lst), max(lst)

lo, hi = min_max([4, 1, 9, 2])
print(f'min={lo}, max={hi}')
```

## 5. Quick Comparison

| Feature | List | Dict | Set | Tuple |
|---------|------|------|-----|-------|
| Ordered | Yes (Python 3.7+ for dict too) | Yes (insertion) | No | Yes |
| Mutable | Yes | Yes | Yes | **No** |
| Duplicates | Yes | Keys: No, Values: Yes | **No** | Yes |
| Indexed | Yes | By key | No | Yes |
| Hashable | No | No | No (frozenset yes) | Yes |
| Syntax | `[...]` | `{k:v}` | `{...}` | `(...)` |
